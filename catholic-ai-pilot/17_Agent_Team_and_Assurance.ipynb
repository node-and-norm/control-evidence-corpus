## Spec-driven engineering: before coding, specify what must remain true

The [versioned specification](_evidence/agent-team/spec.json) was committed at `0205eef` **before implementation**. It fixes roles, allowed capabilities, input/output states, disallowed claims, trace requirements and acceptance tests. The sequence is **specify → plan → implement → test against the spec → review → controlled release**.

[Microsoft’s spec-first guidance](https://developer.microsoft.com/blog/spec-driven-development-ai-native-engineering/) motivates this approach. We are applying the method, not claiming installation or execution of GitHub Spec Kit. New requirements must produce an explicit spec revision and new tests; agents cannot rewrite their permissions.

Software integration is separate from approving a research protocol or source interpretation. [Issue #41](https://github.com/node-and-norm/control-evidence-corpus/issues/41) remains the study-owner gate.

## A bounded team, with accountable humans outside it

| Role | Bounded task |
|---|---|
| **risk incident linker** | Keep risks, alleged incidents, external analogues and synthetic tests separate; link retained IDs only. |
| **source verifier** | Check source identity, edition, paragraph locator and retained hash; semantic support stays a distinct question. |
| **obligations mapper** | Keep civil-law candidates, canonical provisions, Church teaching and proposed engineering controls distinct. |
| **assurance critic** | Identify unsupported conclusions, missing intervention evidence and residual risks; challenge is not independent human review. |
| **synthesis writer** | Produce a bounded review packet with contribution disclosure; never publish or claim approval. |

A deterministic coordinator fixes the order and budget. A canon lawyer, theologian, privacy/technical reviewer and the research owner remain **human roles**, not AI personas. Several models agreeing would not establish independent peer review. The same assistant authored this prototype, its fixtures and its tests.

## The workflow

**Frozen synthetic packet → risk/incident links → source correspondence → obligation candidates → assurance critique → PAUSE → synthetic test acknowledgment → review-only synthesis.**

The pause uses LangGraph `interrupt()` and an in-memory checkpointer; the resumed node has no repeated role calls before the pause. The acknowledgment is explicitly a fixture token, never a human signature. [Official interrupt documentation](https://docs.langchain.com/oss/python/langgraph/interrupts) explains resumption semantics. Durable cross-process recovery, authenticated reviewers, revocation and live adapters remain unimplemented.

The research process remains **Discover → Screen → Review → Reconstruct → Evaluate → Report**. This prototype tests workflow plumbing, not completion of those scientific stages.

## Where MIT, AI incidents, Magisterium and LegiScan belong

| Input | Proper role | Boundary |
|---|---|---|
| MIT risk records | Candidate failure mechanisms and source IDs | A risk is not an observed institutional event |
| Incident reports | Reported events or external analogues | A report is not proof of causation, harm or Catholic applicability |
| Magisterium and original Church publications | Discover passages; verify original text and edition | Retrieval is not qualified interpretation or ecclesial approval |
| Explorer and LegiScan | Legislative discovery, status, dates and governance context | An enacted or proposed bill is not automatically applicable to the case |
| Institutional control evidence | Actions, opportunities, acknowledgments and effects | A written policy or referral suggestion is not proof an intervention worked |

This prototype reads **invented local records only**. It does not connect these services. The [Explorer](https://nodeandnorm.org/governance-map) explicitly distinguishes research relevance from legal applicability; its [LegiScan review record](https://nodeandnorm.org/assets/governance/legiscan-review.json) is a scoped published snapshot, not a fresh nationwide legal determination.

## Does the Explorer cover safety, assurance, evaluation and governance?

**It supports discovery and context. It does not replace the controls, tests, evidence or accountable decisions below.**

| Area | What this prototype supplies | What remains unestablished |
|---|---|---|
| **AI safety** | Synthetic-only input, no network/provider tools, fixed capability boundary and budget, strict IDs and output states. | Real model adversarial testing, privacy review, deployment monitoring and incident response. |
| **AI evaluation** | Requirement-to-test traceability, positive and negative software fixtures, original outcomes retained. | New independently prepared research cases, qualified ratings, matched checklist comparator, coverage and abstention metrics. |
| **AI assurance** | Versioned specification, input hashes, exact visible traces, limitations and review-only outputs. | Independent review, claim-evidence argument and actual operational-control evidence; no blanket certification. |
| **AI governance** | Issue #41 and separate human review/release gates; agents cannot approve themselves. | Authenticated reviewers, applicable obligations, incident/correction process and change control. |
| **Research integrity** | Role-level AI-use objects, exact prompts/results, unknown model details, separate dated publisher checks. | Venue selection and current manuscript-specific reconciliation; human accountability cannot be delegated. |

[NIST AI RMF Core](https://airc.nist.gov/airmf-resources/airmf/5-sec-core/) organizes risk management around Govern, Map, Measure and Manage. This is an organizing crosswalk, not a certification or complete conformance assessment. We must evaluate both **the AI used to conduct our research** and **the institutional AI system being studied**.
