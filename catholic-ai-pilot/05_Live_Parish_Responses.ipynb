## PQA-01

### Exact submitted prompt

Respond in your own words in at most 250 words. Cite sources using document and canon or paragraph references. Paraphrase instead of quoting source passages.

In a Latin Church parish planning an AI information assistant, what does canon 220 protect? Explain the principle without selecting a technical hosting architecture.

### Recorded Magisterium answer

Canon 220 protects two related rights: a person’s good reputation and that person’s privacy. It forbids illegitimate harm to someone’s reputation and recognizes each person’s right to protect private matters (1983 Code of Canon Law, can. 220). [^1]

For a parish AI information assistant, the principle is that using information to serve parishioners does not make every personal detail appropriate to collect, disclose, or repeat. The parish should respect the dignity and legitimate interests of people whose information the assistant may encounter, and avoid disclosures or outputs that needlessly expose private matters or unjustly damage someone’s standing. The canon states the governing right; it does not, by itself, prescribe a particular technical design or hosting arrangement.

## PQA-02

### Exact submitted prompt

Respond in your own words in at most 250 words. Cite sources using document and canon or paragraph references. Paraphrase instead of quoting source passages.

A parish vendor says canon 220 requires a locally hosted AI model and guarantees that locally hosted software cannot leak personal data. Does that conclusion follow from this canon?

### Recorded Magisterium answer

No. **Canon 220 protects a person’s right to privacy; it does not prescribe a particular technology or require a parish to use a locally hosted AI model.** The canon forbids illegitimate injury to a person’s good reputation and violation of the right to protect privacy (1983 Code of Canon Law, can. 220). [^1]

Nor does local hosting guarantee that personal data cannot leak. A 2023 Vatican circular warns that computer and online tools, including highly advanced ones, cannot provide certainty of confidentiality and secrecy, and that their use can create risks to legally protected rights (Dicastery for Institutes of Consecrated Life and the Societies of Apostolic Life, *Circular Letter on the Utilization of Computer-Telematic Tools*, no. 3). [^4]

So the canon supports taking privacy seriously, but the vendor’s two claims go beyond what it says. Whether local hosting is appropriate depends on the system’s actual safeguards and risks; the cited materials do not establish that local hosting is required or leak-proof.

## PQA-03

### Exact submitted prompt

Respond in your own words in at most 250 words. Cite sources using document and canon or paragraph references. Paraphrase instead of quoting source passages.

For a Latin Church parish, which questions do canons 824, 827 and 829 raise about publishing a catechetical resource and its later revised edition? Explain what remains for a competent authority to determine.

### Recorded Magisterium answer

For a Latin Church parish, the first question is whether the resource is a catechism or another writing intended for catechetical instruction. If so, canon 827 §1 requires the local ordinary’s approval for publication; this includes translations. Canon 824 §2 extends the rules to writings intended for public distribution, not only conventionally published books. [^2] [^1]

The parish must also identify the appropriate ordinary. Unless another rule applies, canon 824 §1 points to the local ordinary of the author or of the place where the work is published. If the resource is to be exhibited, sold, or distributed in a church or oratory, canon 827 §4 separately requires permission from, or subsequent approval by, competent ecclesiastical authority. If it will serve as a school instructional text, canon 827 §2 raises an additional approval requirement. [^1] [^2]

A revised edition raises a distinct issue: under canon 829, approval or permission for the original text does not automatically extend to a new edition or translation. The revised edition therefore needs its own authorization where the relevant canon requires one. [^3]

The canons leave the competent authority to determine the proper procedure and whether the proposed use—publication, parish distribution, church sale, or school instruction—engages additional requirements. They do not specify how extensive a revision must be to count as a “new edition,” or resolve in advance which ordinary should act if the author’s and publication locations differ or another rule governs.

## PQA-04

### Exact submitted prompt

Respond in your own words in at most 250 words. Cite sources using document and canon or paragraph references. Paraphrase instead of quoting source passages.

Our parish chatbot cites an approved catechism. May we describe every newly generated answer as carrying that catechism's approval, without further review? Explain the limits of that inference.

### Recorded Magisterium answer

No. Approval of the catechism does not automatically extend to every answer a chatbot generates from it. The approval concerns the catechism itself, not new material produced using it.

Canon 827 §1 requires ecclesiastical approval for the publication of catechisms and other writings pertaining to catechetical instruction. The 1992 CDF instruction explains that approval concerns the particular writing examined: it indicates that the writing contains nothing contrary to the Church’s authentic teaching on faith or morals and that relevant canonical requirements have been met (CDF, *Instruction on Some Aspects of the Use of the Instruments of Social Communication*, II.7 §2). It does not certify unreviewed outputs that may paraphrase, combine, omit, or misrepresent that writing.

Likewise, John Paul II describes the *Catechism of the Catholic Church* as a reliable reference for teaching Catholic doctrine, while distinguishing it from local catechisms that require their own approval (*Fidei Depositum*, IV). The approval of a reference text therefore does not make every later explanation an officially approved catechetical text. Even approval to publish a catechism is distinct from adopting it as the official diocesan text (CDF response in *Acta Apostolicae Sedis* 76 [1984], p. 51).

You may accurately say the chatbot **uses or draws on** an approved catechism. Calling every generated answer “approved by” that catechism would overstate what the approval establishes. Each answer’s accuracy and faithful presentation would need to be checked before claiming ecclesiastical approval for it.

## PQA-05

### Exact submitted prompt

Respond in your own words in at most 250 words. Cite sources using document and canon or paragraph references. Paraphrase instead of quoting source passages.

Explain what canons 965 and 966 say about the minister of penance and the faculty for absolution. Keep this an educational explanation.

### Recorded Magisterium answer

Canon 965 identifies the minister of the sacrament of penance: **only a priest** can administer it. Thus, priestly ordination is necessary; a deacon or layperson cannot validly impart sacramental absolution (1983 Code of Canon Law, can. 965). [^1]

Canon 966 adds a distinct requirement for validity. A priest must have not only the power received through ordination, but also the **faculty** to exercise that power for the faithful person he absolves. Being a priest, by itself, therefore does not suffice for valid absolution in every circumstance (1983 Code of Canon Law, can. 966 §1). [^2]

The faculty may arise **by law** or be granted by the competent authority, as determined by canon 969. The canons thus distinguish the priest’s sacramental capacity, conferred in ordination, from the faculty governing its valid exercise for particular faithful (1983 Code of Canon Law, can. 966 §§1–2). [^2] [^6]
