**Central question:** With identical available evidence, can a structured CEC record help qualified reviewers reach more defensible conclusions than a well-designed uncertainty-aware checklist?

**Research boundary:** This notebook reviews published work and proposes a *future* comparison. It does **not** freeze an experiment, re-score the six observed Magisterium answers, replace independent theologian/canon-law judgment or claim empirical benefit.

All earlier reports, held development cases and original source-response records remain unchanged.


## For Catholic leaders: why this comparison matters

Imagine two reviewers assessing an AI assistant's answer to a parish question. Both see the same Church passages, the same answer and the same record of who could intervene.

One reviewer uses a careful checklist. The other receives a structured record distinguishing the **document**, **meaning**, **institutional authority**, **opportunity to intervene**, and **observed effect**.

Our proposed study asks whether that structure helps the second reviewer make fewer unsupported judgments **without unnecessarily refusing well-supported answers or taking too much additional time**.

A diagram or more fields cannot establish trustworthiness by itself. A simpler method that performs equally well with less burden may be the preferred outcome.

### Owen · 2026

**Publication:** [Owen, J. Detecting doctrinal flattening in AI generated responses. AI and Ethics 6, 178 (2026).](https://link.springer.com/article/10.1007/s43681-026-01051-0)  
**Permanent identifier:** `10.1007/s43681-026-01051-0`  
**Evidence consulted:** `publisher_full_html_article`.

**What was actually studied:** 576 source-derived atomic doctrinal claims from 11 Christian traditions, compared with two model outputs using structured text matching and doctrinal heuristics.

**What overlaps our pilot:** Direct precedent for doctrinal-claim support, omission and tradition-specific theological differences.

**What remains unestablished from the reviewed material:** Human institutional reviewer intervention and effectiveness of documentary control records are not its primary dependent variables.

**Research implication:** Account for omission and qualifiers, not only contradictions; distinguish automated reference labels from independently verified theology.


### FMG · 2026

**Publication:** [Chao, A. When AI Is Your Pastor: A Benchmark for Theological Triage and Pastoral Guidance in Large Language Models. FMG-Bench (2026 preprint).](https://arxiv.org/abs/2608.12324)  
**Permanent identifier:** `10.48550/arXiv.2608.12324`  
**Evidence consulted:** `arxiv_full_html_preprint`.

**What was actually studied:** 120 Christian pastoral/doctrinal scenarios, 14 models and 8,792 scored responses across guided and unguided instruction conditions; measures escalation appropriateness.

**What overlaps our pilot:** Direct precedent for pastoral redirection, authority-sensitive triage and evaluation of prompts and instructions.

### SourceCheckup · 2025

**Publication:** [An automated framework for assessing how well LLMs cite relevant medical references (SourceCheckup). Nature Communications (2025).](https://www.nature.com/articles/s41467-025-58551-6)  
**Permanent identifier:** `10.1038/s41467-025-58551-6`

### O2C · 2026

**Publication:** [Hardy, K. From human oversight to effective control: A comparative documentary framework for high-impact AI systems. Digital Policy (online first 2026).](https://www.accscience.com/journal/DP/articles/online_first/9411)  
**Permanent identifier:** `10.36922/DP026290034`  
**Evidence consulted:** `publisher_abstract_and_metadata`.

**What was actually studied:** Eight prescriptive instruments, 19 public cases and 41 oversight/control evidence statements; six functions Observe, Detect, Interpret, Challenge and Decide, Intervene, Recover and Learn.

## Interpreting the initial 24-case result

Our fixed **CIR-001** engineering comparison used 22 assistant-authored synthetic packets and two known metadata replays. The simple checklist and structured gate each retained **8/8** constructed valid identities and held **16/16** nonadmissible or unresolved cases. The structured gate showed **no admission advantage**.

Every outcome left semantic meaning unreviewed. The same assistant designed the engine and expected cases. No independent human reviewer outcomes, theological correctness or real institutional controls were measured.

**Interpretation:** The next experiment must **earn** a complexity advantage under equal-information conditions. A tie with lower checklist cost or poorer CEC performance is scientifically meaningful and can change the method choice.

## Prospective study: what exactly will be compared?

| Design element | Proposed rule, **not frozen** |
|---|---|
| Unit | One bounded AI-assisted institutional control opportunity, tied to a specific event, source, authority and relevant time window |
| Arm A: strong comparator | A well-designed narrative/checklist that explicitly records sources, support, applicability, authority, intervention, uncertainty and citations |
| Arm B: CEC record | Structured propositions, source assertions, institutional scope, authorization, opportunity, intervention, receipt/effect and missing evidence |
| Information parity | Both arms receive exactly the **same source content and operational records**, time horizon and permitted context |
| Presentation | Different representations of equal evidence. No arm may secretly receive a better source retrieval or more facts |
| Reviewers | Qualified reviewers with recorded competence, conflicts and previous AI exposure; independent initial ratings prior to cross-arm analysis |
| Allocation | Randomize reviewer-case assignment and balance case difficulty; an individual must not see the same case in both arms before the first assessment |
| Outcomes | Unsupported categorical conclusions, retained supported decisions, appropriate uncertainty, agreement, elapsed time and propagation/authority errors |
| Analysis | Predeclare estimand, denominator, missingness, exclusions, abstentions, thresholds, power and limitations **before** running |
| Human approval | Issue #41 remains open; the governing comparative protocol remains **0.3.0 and unfrozen** |

**Scope control:** This is one sequential study with source, authority and operational evidence strata. It is not three parallel new experiments. The initial six Magisterium answers and the authored CIR-001 packets are development material and cannot be called an unseen confirmatory set.


## Candidate outcomes, designed to resist misleading success claims

**Primary candidate, pending owner approval:** The share of independently adjudicated *eligible case decisions* on which a reviewer makes a categorical claim unsupported by the supplied evidence. The denominator, adjudication categories and how indeterminate cases enter it must be fixed before observation.

**Non-negotiable safeguard against trivial abstention:** Also measure retention of **correct determinate conclusions** on cases independently adjudicated as answerable. A process that refuses every case should not be declared superior.

**Secondary candidates:**

| Construct | Required evidence | Failure to recognize |
|---|---|---|
| Uncertainty retention | Reviewer explicitly preserves unknown or conflicting support | Treating unknown as approved or rejected without basis |
| Source support | Original passage and claim-to-passage rationale | Treating an authentic URL as proof of the full claim |
| Institutional applicability | Applicable setting, jurisdiction and competent authority | Extending a source beyond its documented scope |
| Practical human control | Notice, opportunity, decision, action, acknowledgement and effect | Confusing a written referral or approval policy with an executed intervention |
| Reviewer cost | Timed initial assessment and documented clarification steps | Achieving similar outcomes at materially greater burden |
| Agreement | Independent pre-adjudication ratings and preserved dissent | Calling post-discussion consensus independent reliability |

**No numerical success threshold or sample size is specified yet.** These require a research-owner decision and a defensible design calculation. Expert review and permitted access to original evidence are essential before adjudication.


## What would disconfirm our contribution?

1. The uncertainty-aware checklist matches CEC decision quality and requires less time.
2. CEC makes reviewers more confident about weak evidence or hides unknowns behind structure.
3. Any apparent advantage disappears when both arms receive the same complete evidence.
4. Independent reviewers disagree so substantially that no stable source or authority standard can be established.
5. An apparent human intervention lacks evidence that the action reached the responsible person or changed the outcome.

These are potential **research findings**, not implementation problems to conceal. An unsuccessful complex method would still yield a publishable, bounded result if measured and reported rigorously.
