## The rule for interpreting evidence

![Evidence sequence: source identity, meaning, applicability, observed operation, qualified review](assets/evidence-path.svg)

A correct citation is not proof of a sound interpretation. A plausible interpretation is not proof that a control operated. A passing software check is not independent human review.
