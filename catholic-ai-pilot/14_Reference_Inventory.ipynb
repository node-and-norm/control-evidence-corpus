## Audit method, limitations and next gate

**Method:** Load the fixed `PARISH-QA-001-R1/evidence.json` from the repository, enumerate each `response.references` entry in order, preserve its `document_title`, `document_reference`, `document_author` and `source_url`, and group by exact URL and exact title strings. All counts can be independently recomputed from the retained record.

**Evidence identity:** source Git blob `52492d739b7ae669d7bfedfd373f43073b5106f1`. The source-run plan and six answers have not been modified by this task.

**Limitations:** The inventory has no source-content fetches, canonical ratings, doctrinal labels, error-frequency estimates, or claims about provider reliability. Blank authors are missing metadata, not anonymous authorship. URL uniqueness does not establish document uniqueness or passage support. The service's returned bibliography may include supplementary reading not explicitly cited in the answer text.
