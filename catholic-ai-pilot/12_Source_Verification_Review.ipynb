### Provisional source-to-claim review

This document preserves an **AI-assisted, post-outcome** source correspondence audit of the six existing Magisterium responses. All classifications are subject to independently qualified review. It is *not* a formal accuracy evaluation, canonical opinion or institutional compliance determination. No new Magisterium chat calls were made.

**Case population:** PQA-01 through PQA-06, as previously captured. **Research status:** exploratory; qualified review pending. The original responses remain in the linked source notebook and evidence JSON.

## Review method and source hierarchy

1. Identify an individual proposition in each recorded answer.
2. Locate the cited canon, doctrinal paragraph or document in the original issuer's publication when available.
3. Record whether the passage directly supports the proposition, provides contextual support, or leaves an inference or locator issue.
4. Keep institutional applicability, technical effectiveness and review authority separate.
5. Preserve disagreement and unresolved status. Do not revise the original answer or infer a provider accuracy score.

**Evidence limitation:** Review occurred after the six answers were visible, so the propositions and suggested classifications are exploratory and cannot serve as blind reference labels. A later independent reviewer should work from the separate original answers and primary sources before opening this annotated packet.

## Areas requiring particular scrutiny

- **2023 circular (§3):** The confidentiality warning is in the original two-page English text. It addresses governance of institutes of consecrated life under canons 627, 127 and 166; application to general parish-AI hosting is not established.
- **Catechism ¶1421:** The cited paragraph discusses the two healing sacraments. The more direct description of priestly absolution appears at ¶1424. This is a citation-specific support question, not a silent revision of the recorded output.
- **Denzinger 1684/1994:** The numbering is edition-dependent and was corroborated through secondary references only. A qualified reviewer should inspect the correct edition and original historical source.
- **‘Referral’ language:** The model told the fictional requester to approach a priest; no referral was initiated or delivered.

These issues are recorded as research leads, not alleged vendor defects or definitive doctrinal mistakes.

## Next controlled comparison: proposed design, not yet frozen

**Question:** Does a source-and-authority gate reduce unsupported affirmative claims while retaining valid explanations when compared against a simple baseline?

**Required controls:** Same permitted source packet, matched case conditions, documented tool access, held-out scenarios and independent source judgments. Existing PQA-01–06 are *development data* and cannot be reused as an unseen holdout.

**Proposed observations:** unsupported admissions, correct retention, appropriate abstention, false authority, reviewer workload, dissent and all invalid attempts. Distinguish technical results from canonical applicability.

**Freeze gate:** The research owner must approve source population, source rights, reviewer qualifications and independence rules, endpoints, denominator, exclusion policy and stop criteria *before* a new confirmatory run.

**Current decision:** No confirmatory run is authorized or reported by this packet.
