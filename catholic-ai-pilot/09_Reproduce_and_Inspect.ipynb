**These are shell commands that launch Python, not Python source to paste into a standard Python notebook cell.** Keep the fenced command blocks as instructions; reading this page must not install software or run tests automatically.

Use a **Python 3.11** environment on Linux or macOS, matching the validated interpreter version. The trace recorder uses POSIX file locking. For the full before/after comparison, use a **Git clone with its history**, not only a downloaded source ZIP: the comparison reads baseline commit `ab5140f9b80d84380186496c0d89c7f6323de5b6` from Git. Open Terminal in the repository root, where `pyproject.toml` and `CATHOLIC_PILOT.ipynb` are located. In your chosen environment, check the interpreter:

```bash
python --version
```

**Install once before running the checks.** The `.[agents]` argument installs the local runtime and its optional LangGraph dependency; the two requirements files add the existing validation and notebook dependencies.

```bash
python -m pip install '.[agents]' -r requirements.txt \
  -r catholic-ai-pilot/requirements.txt
```

The backslash continues the same command on the next line in a POSIX shell. Package installation can use the internet to obtain dependencies. The following research-data checks do not call Magisterium, LegiScan or an AI model. This setup replaces the earlier first installation block that omitted the installed runtime package.
