## Research question

Can a source-traceable, event-control evidence method improve the support and reproducibility of AI risk and safeguard assessments in Catholic institutional settings, compared with simpler narrative and checklist representations?

## Proposed contribution

A bounded, testable crosswalk connecting general AI failure mechanisms, Catholic institutional workflows, source-located obligations, candidate controls, and what independent reviewers can establish from surviving evidence. The added value must be demonstrated against simpler representations.

### Competing explanation

A source-linked narrative or a clear checklist may achieve comparable reconstruction with much less effort. If so, the additional CEC structure must be narrowed or justified by another measured benefit.

## The decision this study serves

Can a Catholic institution justify a claim that an AI safeguard operated, and can another reviewer follow the evidence supporting that claim? The main study compares a CEC record with a source-linked narrative and a checklist using the same evidence. It remains exploratory and awaits independent reviewers and a frozen comparison design.



## What we are studying first

Pastoral and catechetical AI assistants, and Catholic education. The first activity is a bounded source-verification run: whether a requested Church-document paragraph can be discovered, fetched, and identified without silently repairing a mismatch. This is upstream evidence preparation. It does not test a deployed institution or safeguard.

The research objects stay separate: possible risk, reported incident, source proposition, researcher interpretation, event-control observation, and controlled test. The 18 existing mappings from Antiqua et Nova and Magnifica Humanitas are provisional interpretations requiring qualified review.

Use the first exact-document, exact-paragraph hit. Fetch its exact returned identifier once. Preserve not-found and wrong-paragraph responses. Do not use a previously known identifier as a replacement. Stop after two searches and at most two fetches. Original Vatican checks are a separate verification step.



## What would count as evidence

Retain the issued query, all displayed result metadata, chosen identifier, fetch fields, chronological recording notes, and official paragraph checks. Preserve the plan hash. Where source-text republication rights are unresolved, retain a hash and short excerpt in the public record and explain the omission. Recorded-at timestamps describe local capture after return, not provider timing.



## What this work cannot establish

These checks cannot estimate corpus-wide retrieval reliability, attribute a provider defect, establish doctrinal or canonical validity, count as independent human review, or demonstrate a safeguard's operation. The first five institutional risk scenarios and subsequent comparative and safeguard studies still require source intake and review.



## How we will iterate

Keep the current plan and outputs. A retry, changed query, different identifier, or additional target is a new run or explicitly logged amendment. A negative result is retained. A proposed normative mapping can be rejected without altering the underlying document. Human reviewers retain their initial judgments and disclose AI exposure.
