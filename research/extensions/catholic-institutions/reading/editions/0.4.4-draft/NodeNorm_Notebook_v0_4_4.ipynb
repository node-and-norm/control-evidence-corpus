## The research problem

This study asks a specific question: when a Catholic institution uses AI, what evidence would justify confidence that a relevant safeguard actually worked?

> A documented risk is a reason to investigate. It is not proof that harm occurred. A written safeguard is a claim. Its effectiveness must be supported by evidence.

**Principal question**

Can a source-traceable, event-control evidence method improve the support and reproducibility of AI risk and safeguard assessments in Catholic institutional settings, compared with simpler narrative and checklist representations?

**Contribution to test:** A bounded, testable crosswalk connecting general AI failure mechanisms, Catholic institutional workflows, source-located obligations, candidate controls, and what independent reviewers can establish from surviving evidence. The added value must be demonstrated against simpler representations.

**Competing explanation:** A source-linked narrative or a clear checklist may achieve comparable reconstruction with much less effort. If so, the additional CEC structure must be narrowed or justified by another measured benefit.

## Hypotheses and failure conditions

### H1 — Better-supported reconstruction (Candidate primary)

CEC-style event-control records will reduce unsupported categorical conclusions against an uncertainty-aware narrative and checklist, when all reviewers receive identical source material.

**Measure:** Unsupported yes/no conclusions per applicable control opportunity; also unresolved-state retention, traceability and reviewer burden.

**Challenge:** Comparable outcomes at lower cost for simpler representations would disconfirm practical advantage.

### H2 — Reliable Catholic contextualization (Exploratory)

Independent technical and qualified Catholic reviewers can consistently identify which institutional obligations and workflows are implicated by a documented technical risk.

**Measure:** Independent agreement, retained disagreements, unclassifiable cases and revision costs.

**Challenge:** Unstable categories or interpretive disagreements may make the crosswalk unsuitable for empirical use.

### H3 — Testable safeguard response (Exploratory)

A specified candidate safeguard reduces a frozen failure outcome against a defined baseline within a controlled test setting.

**Measure:** Observed failures, missed benign requests, false positives, invalid responses and uncertainty across conditions.

**Challenge:** Inconsistent, trivial, or context-dependent differences limit assurance claims.

### H4 — What public evidence cannot settle (Descriptive)

Some institutional control questions remain unresolved because reports omit configuration, authority, deployment or downstream records.

**Measure:** Counts of unknown, not reported, not observable and conflicting states under a fixed retrieval procedure.

**Challenge:** Cannot infer actual absence of a safeguard from missing records.

## The boundaries we protect

- **Risk ≠ incident:** A risk record describes a possible failure. Incident evidence concerns a reported event.
- **Source count ≠ independence:** Repeated reporting of one institutional assertion does not create new independent evidence.
- **Policy ≠ operation:** A written safeguard alone does not establish that it affected a decision.
- **Catholic relevance ≠ doctrinal judgment:** Normative applicability needs precise sources and appropriately qualified review.
- **Synthetic ≠ deployed:** Laboratory test outcomes do not establish effectiveness in a parish, school or diocese.
- **Missing evidence ≠ failure:** A report that says nothing about a control cannot prove the control was absent.
- **AI citation ≠ verified original:** A returned passage ID or search link is not accepted until the fetched locator and original Vatican paragraph agree.

A Catholic doctrinal or canonical source does not automatically establish that a particular deployment violated Church teaching; qualified interpretation is separate.
